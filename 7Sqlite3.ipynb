```python
import sqlite3
```

```python
## connect to an SQLite database
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
## Create a table
cursor.execute('''
CREATE TABLE employes (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    department TEXT
)
''')

## commit the changes
connection.commit()
```

```python
cursor.execute('''
Select * from employes
               ''')
```

```python
## Insert data in sqlite database
cursor.execute('''
INSERT INTO employes (id, name, age, department)
VALUES (1, 'Om', 21, 'Data Scientist')
''')

cursor.execute('''
INSERT INTO employes (id, name, age, department)
VALUES (3, 'Aarav', 23, 'ML Engineer')
''')

cursor.execute('''
INSERT INTO employes (id, name, age, department)
VALUES (4, 'Riya', 21, 'Data Analyst')
''')

## connection.commit
connection.commit()
```

```python
## Query the database and fetch all records from the employes table

cursor.execute('Select * from employes')
row=cursor.fetchall()

## print the query data
for row in row:
    print(row)
```

```python
## Update records in the employes table
cursor.execute('''
UPDATE employes
SET age =22
WHERE NAME='Riya'
               ''')

connection.commit()
```

```python
cursor.execute('Select * from employes')
row=cursor.fetchall()

## print the query data
for row in row:
    print(row)
```

```python
cursor.execute('''
DELETE FROM employes
WHERE name='Aarav'
               ''')

connection.commit()
```

```python
# import sqlite3

## Connection to an SQLite database
connection = sqlite3.connect('sales_db')
cursor = connection.cursor()

## Create a table for sales data
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales (
  id INTEGER PRIMARY KEY,
  date TEXT NOT NULL,
  product TEXT NOT NULL,
  sales INTEGER,
  region TEXT
)
''')

## Insert sample sales data
sales_data = [
  ('2023-01-01', 'Product A', 100, 'North'),
  ('2023-01-02', 'Product B', 150, 'South'),
  ('2023-01-03', 'Product A', 200, 'East'),
  ('2023-01-04', 'Product C', 250, 'West'),
  ('2023-01-05', 'Product B', 300, 'North')
]

cursor.executemany('''
INSERT INTO sales (date, product, sales, region)
VALUES (?, ?, ?, ?)
''', sales_data)

connection.commit()

```

```python
cursor.execute('Select * from sales')
row=cursor.fetchall()

## print the query data
for row in row:
    print(row)
```

```python
## close the conneection
connection.close()
```
