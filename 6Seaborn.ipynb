```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
## Basic piotting with seaborn
tips=sns.load_dataset('tips')
tips
```

```python
## create a scatter plot
sns.scatterplot(x='total_bill', y='tip' ,data =tips)
plt.title('Scatter plot of total bill vs tip')
plt.show()
```

```python
## Line plot
sns.lineplot(x='size',y='total_bill', data=tips)
plt.title('Line plot of size vs total bill')
plt.show()
```

```python
## categorial plot
sns.barplot(x='day', y='total_bill', data=tips)
plt.title('Bar plot of total bill by day')
plt.show()
```

```python
# Box plot
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Box plot of total bill by day')
plt.show()
```

```python
## violin plot
sns.violinplot(x='day', y='total_bill', data=tips)
plt.title('Volin plot of total bill by day')
plt.show()
```

```python
## Histogram plot
sns.histplot(tips['total_bill'],bins=10, kde=True)
plt.title('Histogram of total bill')
plt.show()
```

```python
## kde plot
sns.kdeplot(tips['total_bill'], fill=True)
```

```python
## pair plot
sns.pairplot(tips)
```

```python
corr=tips[['total_bill', 'tip', 'size']].corr()
corr
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
```

```python
## Style & Aesthetics
sns.set_style("whitegrid")  # options: white, dark, whitegrid, darkgrid, ticks



sns.set_palette("Set2")  # try: deep, Set1, coolwarm, husl, etc.
sns.scatterplot(x="total_bill", y="tip", data=tips, hue="sex")
plt.show()

sns.set_context("talk")  # options: paper, notebook, talk, poster
sns.boxplot(x="day", y="total_bill", data=tips)
plt.title("Boxplot with Presentation Style")
plt.show()
```
