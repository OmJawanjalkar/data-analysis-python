```python
import logging

## configure the basic logging setting
logging.basicConfig(level=logging.DEBUG)

# Log messages with different severity levels
logging.debug("This is a debug message")     # For detailed debugging info
logging.info("This is an info message")      # General info
logging.warning("This is a warning")         # Something to be cautious about
logging.error("This is an error")            # An error occurred
logging.critical("This is critical")         # Serious error
```

```python
import logging

# Configuring logging
logging.basicConfig(
    filename='app.log',
    filemode='W'
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

# Log messages with different severity levels
logging.debug("This is a debug message")     # For detailed debugging info
logging.info("This is an info message")      # General info
logging.warning("This is a warning")         # Something to be cautious about
logging.error("This is an error")            # An error occurred
logging.critical("This is critical")         # Serious error
```
